## Project 0: Is a Hot Dog a Sandwich?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Hotdog Survey (Responses) - Form Responses 1.csv')
df.head(20)
```

```python
# Check Missing Values
plt.figure(figsize=(8,5))
sns.heatmap(df.isnull(), cmap='plasma', cbar=False, yticklabels=False)
plt.title("Missing Values in the Dataset")
plt.show()
```

```python
# Remove Graduate Students and Other/Not Affiliated
df = df[~df['Year at UVA'].isin(['Graduate Student', 'Other/Not Affiliated'])]
df.head(20)
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Is a Hotdog a Sandwich?", data=df, palette="pastel")
plt.title("Is a Hotdog a Sandwich?")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Year at UVA", data=df, palette="pastel")
plt.title("Number of students from each year")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="What Schools are you a Part of?", data=df, palette="pastel")
plt.title("School students are a part of")
plt.show()
```
