```python
import numpy as np
import pandas as pd
```

```python
hotdog = pd.read_csv('Hotdog Survey.csv')

hotdog.head()
```

```python
# rename columns for easier access
names = ['timestamp','year','gender','hotdog_opinion','opinion_strength','consumption_freq']
hotdog.rename(columns={'Timestamp': 'timestamp',
                       'Year at UVA': 'year',
                       'What is your Gender Identity?': 'gender',
                       'Is a Hotdog a Sandwich?': 'hotdog_opinion',
                       'How strongly do you feel about your previous answer?': 'opinion_strength',
                       'How frequently do you eat hot dogs?': 'consumption_freq'}, inplace=True)

hotdog.head()
```

```python
# performs one-sample z-test
# H0: p = 0.25 vs H1: p != 0.25

p0 = 0.25  # null hypothesis proportion
n = hotdog.shape[0]  # sample size
SE = np.sqrt(p0 * (1 - p0) / n)  # standard error
p_hat = (hotdog['hotdog_opinion'] == 'Yes').mean()  # sample proportion
z = (p_hat - p0) / SE  # z-score

# find p-value from z-score for two-tailed test
import scipy.stats as stats
p_value = 2 * (1 - stats.norm.cdf(abs(z)))
print(f'Z-score: {z}, P-value: {p_value}')
```

p < 0.05 therefore null hypothesis is rejected and alternative hypothesis is accepted. The proportion of UVA undergraduate students who believe a hotdog is a sandwich is not 25% with 95% significance.
